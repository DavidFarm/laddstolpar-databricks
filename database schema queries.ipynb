```python
%sql
SELECT c.table_schema, c.table_name, t.table_type,
       c.ordinal_position, c.column_name, c.full_data_type, c.is_nullable
FROM laddstolpar_df.information_schema.columns c
JOIN laddstolpar_df.information_schema.tables  t
  USING (table_catalog, table_schema, table_name)
WHERE c.table_schema IN ('landing', 'bronze', 'silver', 'gold', 'ops')
ORDER BY c.table_schema, c.table_name, c.ordinal_position;
```
